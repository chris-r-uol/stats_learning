## Before you start![The five steps of the probability interval method](../assets/img/method-probability.svg)Two ideas are used throughout this tutorial.**A probability is an area.** The probability that a value falls between twolimits is the area under the curve between them. The total area is 1, becauseevery possible value lies somewhere under the curve.**Excel measures area from the left.** `NORM.DIST`, and the cumulative formsof `BINOM.DIST` and `POISSON.DIST`, all return the area from the far left upto your value. If a question asks for a different area, rewrite it using:P(X > c) = 1 − P(X ≤ c)The [reference page](https://chris-r-uol.github.io/stats_learning/reference.html) lists the functions and the area eachone returns. The [symbols glossary](https://chris-r-uol.github.io/stats_learning/reference.html#symbols) explains thenotation.
