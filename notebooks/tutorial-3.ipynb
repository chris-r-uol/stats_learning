## Before you start![The five steps of the interval estimation method](../assets/img/method-intervals.svg)**Use the standard error, not the standard deviation.**A confidence interval does not describe how spread out your data is. Itdescribes how precisely you have measured the **mean**. These are differentquestions with different answers.The spread value you put into `NORM.INV` is the **standard error**. That isthe standard deviation divided by √n.More data gives a narrower interval, because √n grows and the standard errorgets smaller.### Which case applies?Two questions decide the case:1. **Do you know the population standard deviation?** If yes, use σ. If no,   estimate it from the sample and use S.2. **Can you use the normal distribution?** You can if the population is   normal, or if the sample is large enough for the central limit theorem. If   neither is true, use the t-distribution.The table below lists the combinations. The questions refer to these casenumbers.| Case | Population normal? | Population variance known? | Sample size | Use | Statistic ||---|---|---|---|---|---|| **1** | Yes | Yes | any | Inverse Normal | (X̄ − μ) ÷ (σ/√n) ~ Nor(0,1) || **2** | No | Yes | n ≥ 20 | Inverse Normal | (X̄ − μ) ÷ (σ/√n) ~ Nor(0,1) || **3** | Yes | No | any | Inverse t | (X̄ − μ) ÷ (S/√n) ~ t(n−1) || **4** | No | No | n ≥ 30 | Inverse Normal | (X̄ − μ) ÷ (S/√n) ~ Nor(0,1) || **4** | Roughly normal | No | n ≥ 20 | Inverse t | (X̄ − μ) ÷ (S/√n) ~ t(n−1) |
